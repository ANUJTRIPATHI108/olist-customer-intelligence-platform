```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# chart style settings
sns.set_theme(style='whitegrid', palette='muted')
plt.rcParams['figure.dpi']       = 150
plt.rcParams['figure.figsize']   = (10, 5)
plt.rcParams['font.size']        = 11

print("Libraries ready")
```

```python
from sqlalchemy import create_engine
from sqlalchemy.engine.url import URL

connection_url = URL.create(
    drivername = 'postgresql+psycopg2',
    username   = 'postgres',
    password   = 'Anuj@0306',
    host       = 'localhost',
    port       = 5432,
    database   = 'ecommerce_analytics'
)
engine = create_engine(connection_url)

# reload all tables
customers   = pd.read_sql('SELECT * FROM customers', engine)
orders      = pd.read_sql('SELECT * FROM orders', engine)
order_items = pd.read_sql('SELECT * FROM order_items', engine)
payments    = pd.read_sql('SELECT * FROM order_payments', engine)
reviews     = pd.read_sql('SELECT * FROM order_reviews', engine)
products    = pd.read_sql('SELECT * FROM products', engine)
sellers     = pd.read_sql('SELECT * FROM sellers', engine)
translation = pd.read_sql('SELECT * FROM product_category_name_translation', engine)

# fix timestamps
timestamp_cols = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]
for col in timestamp_cols:
    orders[col] = pd.to_datetime(orders[col])

# feature engineering
orders['delivery_delay_days'] = (
    orders['order_delivered_customer_date'] -
    orders['order_estimated_delivery_date']
).dt.days

orders['actual_delivery_days'] = (
    orders['order_delivered_customer_date'] -
    orders['order_purchase_timestamp']
).dt.days

orders['is_late']         = (orders['delivery_delay_days'] > 0).astype(int)
orders['purchase_month']  = orders['order_purchase_timestamp'].dt.to_period('M')
orders['purchase_year']   = orders['order_purchase_timestamp'].dt.year
orders['purchase_hour']   = orders['order_purchase_timestamp'].dt.hour
orders['purchase_dow']    = orders['order_purchase_timestamp'].dt.day_name()

print("Data loaded and ready")
```

```python
status = orders['order_status'].value_counts()

fig, ax = plt.subplots(figsize=(12, 6))
bars = ax.bar(status.index, status.values,
              color=sns.color_palette('muted', len(status)))

# manual labels since bar_label not available in older matplotlib
for bar, val in zip(bars, status.values):
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 200,
        f'{val:,}',
        ha='center', va='bottom', fontsize=9
    )

ax.set_title('Order Status Distribution', fontsize=14, fontweight='bold')
ax.set_xlabel('Order Status')
ax.set_ylabel('Number of Orders')
ax.set_yscale('log')  # log scale so small bars are visible
plt.xticks(rotation=20)
plt.tight_layout()
plt.savefig('chart1_order_status.png', bbox_inches='tight')
plt.show()
print("Chart 1 saved")
```

```python
df_rev = (order_items
          .merge(orders[['order_id',
                          'order_purchase_timestamp',
                          'order_status']], on='order_id')
          .query("order_status == 'delivered'"))

df_rev['month'] = df_rev['order_purchase_timestamp'].dt.to_period('M')
monthly = df_rev.groupby('month')['price'].sum().reset_index()
monthly['month_str'] = monthly['month'].astype(str)

fig, ax = plt.subplots(figsize=(12, 5))
ax.fill_between(monthly['month_str'], monthly['price'], alpha=0.2, color='steelblue')
ax.plot(monthly['month_str'], monthly['price'],
        marker='o', markersize=5, color='steelblue', linewidth=2)
ax.set_title('Monthly Revenue Trend', fontsize=14, fontweight='bold')
ax.set_xlabel('Month')
ax.set_ylabel('Revenue (BRL)')
ax.yaxis.set_major_formatter(mtick.FuncFormatter(lambda x, _: f'R${x:,.0f}'))
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('chart2_monthly_revenue.png', bbox_inches='tight')
plt.show()
print("Chart 2 saved")
```

```python
df_cat = (order_items
          .merge(products[['product_id', 'product_category_name']], on='product_id')
          .merge(translation, on='product_category_name', how='left'))

df_cat['category'] = df_cat['product_category_name_english'].fillna(
                     df_cat['product_category_name'])

top_cat = (df_cat.groupby('category')['price']
           .sum()
           .nlargest(10)
           .reset_index()
           .sort_values('price'))

fig, ax = plt.subplots(figsize=(10, 6))
bars = ax.barh(top_cat['category'], top_cat['price'],
               color=sns.color_palette('Blues_d', 10))
for bar in bars:
    width = bar.get_width()
    ax.text(
        width,
        bar.get_y() + bar.get_height()/2,
        f'R${width:,.0f}',
        va='center',
        fontsize=9
    )
ax.set_title('Top 10 Categories by Revenue', fontsize=14, fontweight='bold')
ax.set_xlabel('Revenue (BRL)')
plt.tight_layout()
plt.savefig('chart3_top_categories.png', bbox_inches='tight')
plt.show()
print("Chart 3 saved")
```

```python
# Payment Type Analysis

pay = payments['payment_type'].value_counts()

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Donut chart
wedges, texts, autotexts = axes[0].pie(
    pay.values,
    autopct=lambda p: f'{p:.1f}%' if p >= 1 else '',
    startangle=90,
    colors=sns.color_palette('muted', len(pay)),
    wedgeprops={'width': 0.45},
    pctdistance=0.75
)

axes[0].legend(
    wedges,
    pay.index,
    title='Payment Type',
    loc='center left',
    bbox_to_anchor=(1.0, 0.5)
)

axes[0].set_title(
    'Payment Type — Transaction Count',
    fontweight='bold'
)

# Revenue by payment type
pay_rev = (
    payments.groupby('payment_type')['payment_value']
    .sum()
    .sort_values(ascending=False)
)

axes[1].bar(
    pay_rev.index,
    pay_rev.values,
    color=sns.color_palette('muted', len(pay_rev))
)

axes[1].set_title(
    'Payment Type — Total Revenue',
    fontweight='bold'
)

axes[1].set_ylabel('Total Revenue (BRL)')

axes[1].yaxis.set_major_formatter(
    mtick.FuncFormatter(lambda x, _: f'R${x:,.0f}')
)

axes[1].tick_params(axis='x', rotation=20)

plt.tight_layout()
plt.savefig('chart4_payment_types.png', dpi=300, bbox_inches='tight')
plt.show()

print("Chart 4 saved")
```

```python
score = reviews['review_score'].value_counts().sort_index()
colors = ['#d73027','#f46d43','#fdae61','#a6d96a','#1a9641']

fig, ax = plt.subplots(figsize=(8, 5))
bars = ax.bar(score.index, score.values, color=colors)
for bar in bars:
    width = bar.get_width()
    ax.text(
        width,
        bar.get_y() + bar.get_height()/2,
        f'R${width:,.0f}',
        va='center',
        fontsize=9
    )
ax.set_title('Review Score Distribution', fontsize=14, fontweight='bold')
ax.set_xlabel('Review Score (1=Worst, 5=Best)')
ax.set_ylabel('Number of Reviews')
ax.set_xticks([1, 2, 3, 4, 5])
plt.tight_layout()
plt.savefig('chart5_review_scores.png', bbox_inches='tight')
plt.show()
print("Chart 5 saved")
```

```python
orders['hour'] = orders['order_purchase_timestamp'].dt.hour
orders['dow']  = orders['order_purchase_timestamp'].dt.strftime('%a')

dow_order = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
pivot = orders.pivot_table(index='dow', columns='hour',
                            values='order_id', aggfunc='count')
pivot = pivot.reindex(dow_order)

fig, ax = plt.subplots(figsize=(14, 5))
sns.heatmap(pivot, cmap='YlOrRd', ax=ax,
            linewidths=0.3, fmt='.0f')
ax.set_title('Orders by Hour and Day of Week',
             fontsize=14, fontweight='bold')
ax.set_xlabel('Hour of Day')
ax.set_ylabel('Day of Week')
plt.tight_layout()
plt.savefig('chart6_heatmap_orders.png', bbox_inches='tight')
plt.show()
print("Chart 6 saved")
```

```python
df_delay = (orders[['order_id', 'delivery_delay_days']]
            .merge(reviews[['order_id', 'review_score']], on='order_id')
            .dropna(subset=['delivery_delay_days']))

df_delay['delay_bucket'] = pd.cut(
    df_delay['delivery_delay_days'],
    bins=[-999, 0, 3, 7, 14, 999],
    labels=['On Time', '1–3d Late', '4–7d Late', '8–14d Late', '15d+ Late']
)

fig, ax = plt.subplots(figsize=(10, 5))
sns.boxplot(data=df_delay, x='delay_bucket',
            y='review_score', palette='RdYlGn_r', ax=ax)
ax.set_title('Review Score vs Delivery Delay',
             fontsize=14, fontweight='bold')
ax.set_xlabel('Delivery Status')
ax.set_ylabel('Review Score')
plt.tight_layout()
plt.savefig('chart7_delay_vs_review.png', bbox_inches='tight')
plt.show()
print("Chart 7 saved")
```

```python
df_state = (customers[['customer_id', 'customer_state']]
            .merge(orders[['order_id', 'customer_id', 'order_status']], on='customer_id')
            .merge(order_items[['order_id', 'price']], on='order_id')
            .query("order_status == 'delivered'"))

state_rev = (df_state.groupby('customer_state')['price']
             .sum()
             .nlargest(10)
             .reset_index()
             .sort_values('price'))

fig, ax = plt.subplots(figsize=(10, 6))
bars = ax.barh(state_rev['customer_state'], state_rev['price'],
               color=sns.color_palette('Greens_d', 10))
for bar in bars:
    width = bar.get_width()
    ax.text(
        width,
        bar.get_y() + bar.get_height()/2,
        f'R${width:,.0f}',
        va='center',
        fontsize=9
    )
ax.set_title('Top 10 States by Revenue', fontsize=14, fontweight='bold')
ax.set_xlabel('Revenue (BRL)')
plt.tight_layout()
plt.savefig('chart8_top_states.png', bbox_inches='tight')
plt.show()
print("Chart 8 saved")
```

```python
order_counts = (customers
                .merge(orders[['order_id', 'customer_id']], on='customer_id')
                .groupby('customer_unique_id')['order_id']
                .count()
                .reset_index())

order_counts['type'] = order_counts['order_id'].apply(
    lambda x: 'Repeat Customer' if x > 1 else 'One-Time Customer'
)

type_counts = order_counts['type'].value_counts()

fig, ax = plt.subplots(figsize=(7, 7))
ax.pie(type_counts.values, labels=type_counts.index,
       autopct='%1.1f%%', startangle=90,
       colors=['#2196F3', '#FF9800'])
ax.set_title('One-Time vs Repeat Customers',
             fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('chart9_repeat_customers.png', bbox_inches='tight')
plt.show()
print("Chart 9 saved")
```

```python
monthly_orders = (orders
                  .query("order_status == 'delivered'")
                  .groupby('purchase_month')['order_id']
                  .count()
                  .reset_index())
monthly_orders['month_str'] = monthly_orders['purchase_month'].astype(str)

fig, ax = plt.subplots(figsize=(12, 5))
ax.bar(monthly_orders['month_str'], monthly_orders['order_id'],
       color='steelblue', alpha=0.8)
ax.set_title('Monthly Order Volume', fontsize=14, fontweight='bold')
ax.set_xlabel('Month')
ax.set_ylabel('Number of Orders')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('chart10_monthly_orders.png', bbox_inches='tight')
plt.show()
print("Chart 10 saved")
```
