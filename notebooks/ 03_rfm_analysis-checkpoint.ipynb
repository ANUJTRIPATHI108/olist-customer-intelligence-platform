```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style='whitegrid', palette='muted')
plt.rcParams['figure.dpi']     = 150
plt.rcParams['figure.figsize'] = (10, 5)

print("Libraries ready")
```

```python
from sqlalchemy import create_engine
from sqlalchemy.engine.url import URL

connection_url = URL.create(
    drivername = 'postgresql+psycopg2',
    username   = 'postgres',
    password   = 'Anuj@0306',
    host       = 'localhost',
    port       = 5432,
    database   = 'ecommerce_analytics'
)
engine = create_engine(connection_url)

customers   = pd.read_sql('SELECT * FROM customers', engine)
orders      = pd.read_sql('SELECT * FROM orders', engine)
order_items = pd.read_sql('SELECT * FROM order_items', engine)

# fix timestamps
orders['order_purchase_timestamp'] = pd.to_datetime(
    orders['order_purchase_timestamp'])

print("Data loaded successfully")
print(f"Customers : {customers.shape[0]:,}")
print(f"Orders    : {orders.shape[0]:,}")
print(f"Items     : {order_items.shape[0]:,}")
```

```python
from IPython.display import display, Markdown

display(Markdown("## RFM Base Dataset"))

df_rfm = (
    customers
    .merge(
        orders[
            ['order_id', 'customer_id',
             'order_purchase_timestamp',
             'order_status']
        ],
        on='customer_id'
    )
    .merge(
        order_items[['order_id', 'price']],
        on='order_id'
    )
    .query("order_status == 'delivered'")
)

display(pd.DataFrame({
    'Metric': ['Rows', 'Columns'],
    'Value': [f"{df_rfm.shape[0]:,}", df_rfm.shape[1]]
}))

display(df_rfm.head(3))
```

```python
# snapshot date = last order date in dataset
snapshot_date = df_rfm['order_purchase_timestamp'].max()
print(f"Snapshot date : {snapshot_date.date()}")

# calculate rfm per unique customer
rfm = (df_rfm
       .groupby('customer_unique_id')
       .agg(
           last_purchase = ('order_purchase_timestamp', 'max'),
           frequency     = ('order_id', 'nunique'),
           monetary      = ('price', 'sum')
       )
       .reset_index())

# recency = days since last purchase
rfm['recency'] = (snapshot_date - rfm['last_purchase']).dt.days

# reorder columns
rfm = rfm[['customer_unique_id', 'recency', 'frequency', 'monetary']]

print(f"\nRFM table shape : {rfm.shape}")
print(rfm.describe().round(2))
```

```python
# score each dimension from 1 to 4 using quartiles
# recency  : lower days = better = higher score
# frequency: higher orders = better = higher score
# monetary : higher spend = better = higher score

rfm['r_score'] = pd.qcut(rfm['recency'],
                          q=4,
                          labels=[4, 3, 2, 1]).astype(int)

rfm['f_score'] = pd.qcut(rfm['frequency'].rank(method='first'),
                          q=4,
                          labels=[1, 2, 3, 4]).astype(int)

rfm['m_score'] = pd.qcut(rfm['monetary'].rank(method='first'),
                          q=4,
                          labels=[1, 2, 3, 4]).astype(int)

rfm['rfm_total'] = rfm['r_score'] + rfm['f_score'] + rfm['m_score']
rfm['rfm_code']  = (rfm['r_score'].astype(str) +
                    rfm['f_score'].astype(str) +
                    rfm['m_score'].astype(str))

print("Scores assigned")
print(rfm[['customer_unique_id', 'recency', 'frequency',
           'monetary', 'r_score', 'f_score',
           'm_score', 'rfm_total']].head(10))
```

```python
def assign_segment(row):
    if row['rfm_total'] >= 11:
        return 'Champions'
    elif row['rfm_total'] >= 9:
        return 'Loyal Customers'
    elif row['rfm_total'] >= 7:
        return 'Potential Loyalists'
    elif row['r_score'] >= 3 and row['f_score'] <= 2:
        return 'New Customers'
    elif row['r_score'] <= 2 and row['f_score'] >= 3:
        return 'At Risk'
    elif row['r_score'] == 1 and row['f_score'] == 1:
        return 'Lost Customers'
    else:
        return 'Need Attention'

rfm['segment'] = rfm.apply(assign_segment, axis=1)

# segment summary
seg_summary = (rfm.groupby('segment')
               .agg(
                   num_customers = ('customer_unique_id', 'count'),
                   avg_recency   = ('recency', 'mean'),
                   avg_frequency = ('frequency', 'mean'),
                   avg_monetary  = ('monetary', 'mean')
               )
               .round(2)
               .sort_values('avg_monetary', ascending=False)
               .reset_index())

print("\nRFM Segment Summary")
print("=" * 70)
print(seg_summary.to_string(index=False))
```

```python
seg_count = rfm['segment'].value_counts()

fig, ax = plt.subplots(figsize=(10, 6))
colors = sns.color_palette('Set2', len(seg_count))
bars = ax.barh(seg_count.index, seg_count.values, color=colors)

for bar, val in zip(bars, seg_count.values):
    ax.text(
        bar.get_width() + 100,
        bar.get_y() + bar.get_height() / 2,
        f'{val:,}',
        ha='left', va='center', fontsize=10
    )

ax.set_title('Customer Count by RFM Segment',
             fontsize=14, fontweight='bold')
ax.set_xlabel('Number of Customers')
plt.tight_layout()
plt.savefig('rfm_chart1_segment_count.png', bbox_inches='tight')
plt.show()
print("Chart 1 saved")
```

```python
seg_rev = (rfm.groupby('segment')['monetary']
           .mean()
           .sort_values()
           .reset_index())

fig, ax = plt.subplots(figsize=(10, 6))
bars = ax.barh(seg_rev['segment'], seg_rev['monetary'],
               color=sns.color_palette('Blues_d', len(seg_rev)))

for bar, val in zip(bars, seg_rev['monetary']):
    ax.text(
        bar.get_width() + 10,
        bar.get_y() + bar.get_height() / 2,
        f'R${val:,.0f}',
        ha='left', va='center', fontsize=10
    )

ax.set_title('Average Revenue by RFM Segment',
             fontsize=14, fontweight='bold')
ax.set_xlabel('Average Revenue (BRL)')
plt.tight_layout()
plt.savefig('rfm_chart2_avg_revenue.png', bbox_inches='tight')
plt.show()
print("Chart 2 saved")
```

## Key RFM Insights

| Segment | Action |
|---|---|
| **Champions** | Reward them — they are your best customers |
| **Loyal Customers** | Upsell higher value products |
| **Potential Loyalists** | Offer membership or loyalty program |
| **New Customers** | Provide onboarding offers |
| **At Risk** | Send win-back campaigns immediately |
| **Lost Customers** | Try reactivation with heavy discount |
| **Need Attention** | Send personalised recommendations |
